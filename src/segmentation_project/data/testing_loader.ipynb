```python
import sys
from pathlib import Path

sys.path.append(str(Path.cwd().parents[2] / "src"))

import matplotlib.pyplot as plt

from segmentation_project.data.loader import ImageLoader
```

```python
# Helper function to find project root (where pyproject.toml is located)


def find_project_root(marker="pyproject.toml"):
    path = Path.cwd()
    while path != path.parent:
        if (path / marker).exists():
            return path
        path = path.parent
    raise FileNotFoundError(f"Project root not found (no {marker})")


project_root = find_project_root()
```

```python
# List all images
images_dir = project_root / "images"
image_paths = ImageLoader.list_images(images_dir)


print("Found images:")
for path in image_paths[:3]:
    print(path)
# Count number of images found
print(f"\nTotal images found: {len(image_paths)}")

# Load the first image and display its shape

image = ImageLoader.load_image(image_paths[0])

print("\nImage loaded successfully")
print("Shape:", image.shape)
```

```python
# Display the first image
plt.imshow(image)
plt.title(image_paths[0].name)
plt.axis("off")
plt.show()
```
