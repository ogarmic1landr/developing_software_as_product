```python
from pathlib import Path

import matplotlib.pyplot as plt

from segmentation_project.data.loader import ImageLoader
```

```python
# List all images
loader = ImageLoader()
image_paths = loader.list_images()

index = 0

# Selected image path
image_path = image_paths[index]

# Load image
image = loader.load_image(image_path)


# Automatically find corresponding mask
mask_path = loader.get_mask_path(image_path)

# Load mask
mask = loader.load_mask(mask_path)


print("Found images:")
for path in image_paths[:3]:
    print(path)
# Count number of images found
print(f"\nTotal images found: {len(image_paths)}")
print("\nImage loaded successfully")
print("Shape:", image.shape)
```

```python
# Display loaded image

plt.imshow(image)
plt.title(image_paths[index].name)
plt.axis("off")
plt.show()
```

```python
# Display Manual mask image

plt.imshow(mask, cmap="gray")
plt.title(mask_path.name)
plt.axis("off")
plt.show()
```
