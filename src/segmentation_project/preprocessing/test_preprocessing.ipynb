```python
from pathlib import Path

import matplotlib.pyplot as plt

from segmentation_project.data import ImageLoader
from segmentation_project.preprocessing import (
    CLAHE,
    Grayscale,
    Resize,
    RGBConverter,
    Sharpen,
)
```

```python
# Load all image paths
loader = ImageLoader()
image_paths = loader.list_images()

# Load first and last image
first_image = ImageLoader.load_image(image_paths[0])
last_image = ImageLoader.load_image(image_paths[-1])
```

```python
# Resize images
first_resized = Resize.apply(first_image)
last_resized = Resize.apply(last_image)

# Display images side by side
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].imshow(first_resized)
axes[0].set_title("First Resized Image")
axes[0].axis("off")

axes[1].imshow(last_resized)
axes[1].set_title("Last Resized Image")
axes[1].axis("off")

plt.tight_layout()
plt.show()
```

```python
# opencv loads images in BGR format, so we need to convert it to RGB for displaying with matplotlib

first_rgb = RGBConverter.apply(first_image)
last_rgb = RGBConverter.apply(last_image)

# Display images side by side
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].imshow(first_rgb)
axes[0].set_title("First Image")
axes[0].axis("off")

axes[1].imshow(last_rgb)
axes[1].set_title("Last Image")
axes[1].axis("off")

plt.tight_layout()
plt.show()
```

```python
# Convert to grayscale

gray_first = Grayscale.apply(first_resized)
gray_last = Grayscale.apply(last_resized)

# Verify dimensions
print("First grayscale shape:", gray_first.shape)
print("Last grayscale shape:", gray_last.shape)


# Display images side by side
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].imshow(gray_first, cmap="gray")
axes[0].set_title("First Image")
axes[0].axis("off")

axes[1].imshow(gray_last, cmap="gray")
axes[1].set_title("Last Image")
axes[1].axis("off")

plt.tight_layout()
plt.show()
```

```python
# sharpen can be applied to any type of image or image output from the previous preprocessing steps

sharp_first = Sharpen.apply(first_rgb)
sharp_last = Sharpen.apply(last_rgb)  # here you can change what image you want to sharpen

# Display images side by side
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].imshow(sharp_first)
axes[0].set_title("First Image")
axes[0].axis("off")

axes[1].imshow(sharp_last)
axes[1].set_title("Last Image")
axes[1].axis("off")

plt.tight_layout()
plt.show()
```

```python
# CLAHE expects only grayscale image, colour image will throw an error

enhanced_first = CLAHE.apply(gray_first)
enhanced_last = CLAHE.apply(gray_last)  # here you can change what image you want to enhance

# Display images side by side
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].imshow(enhanced_first, cmap="gray")
axes[0].set_title("First Image")
axes[0].axis("off")

axes[1].imshow(enhanced_last, cmap="gray")
axes[1].set_title("Last Image")
axes[1].axis("off")

plt.tight_layout()
plt.show()
```
